### 解題思路

#### 觀察圖形
- 從 **(0,0) → (4,4)** 這一段是斜率 = 1 的直線，所以方程式是：
  $$
  y = x
  $$
- 從 **(4,4) → (6,0)** 這一段是往下的斜線，斜率比較陡，計算後發現是：
  $$
  y = -2x + 12
  $$

所以整個函數是 **分段函數 (piecewise function)**：

$$
f(x) =
\begin{cases}
x, & 0 \leq x \leq 4 \\
-2x + 12, & 4 < x \leq 6
\end{cases}
$$

#### 實作策略
1. 先不用一次畫完，而是利用 **for 迴圈跑 x 值**。  
2. 再用 **if 判斷 x 在哪一段**，計算對應的 y。  
3. 把所有 \((x,y)\) 存進 list，最後用 `matplotlib` 畫出來。



```python
import matplotlib.pyplot as plt

x_vals = []
y_vals = []

for x in range(0, 7):  # x從0到6
    if x <= 4:
        y = x
    else:
        y = -2 * x + 12
    x_vals.append(x)
    y_vals.append(y)

plt.plot(x_vals, y_vals, color="orange", linewidth=3)
plt.axhline(0, color="black")  # x軸
plt.axvline(0, color="black")  # y軸
plt.grid(True)
plt.show()
```

### 挑戰題解題思路

#### 觀察圖形
半圓的數學公式可以從圓的參數方程式推導出來。  
若圓心為 $(a, b)$，半徑為 $r$，則完整圓的參數式為：

$$
x = a + r \cos \theta, \quad y = b + r \sin \theta, \quad \theta \in [0, 2\pi]
$$

只要限制 $\theta$ 的範圍，就能得到半圓：
- **上半圓**：$\theta \in [0, \pi]$  
- **下半圓**：$\theta \in [\pi, 2\pi]$  
- **左半圓**：$\theta \in [\pi/2, 3\pi/2]$  
- **右半圓**：$\theta \in [-\pi/2, \pi/2]$  



#### 實作策略
1. 先利用 `numpy.linspace` 產生 $\theta$ 的範圍。  
2. 根據方向（上/下/左/右）決定 $\theta$ 的起訖範圍。  
3. 代入公式 $x = a + r\cos\theta$、$y = b + r\sin\theta$。  
4. 最後用 `matplotlib` 畫出半圓。  



```python
import matplotlib.pyplot as plt
import numpy as np

r = 2  # 半徑
theta = np.linspace(np.pi, 2*np.pi, 200)  # 上半圓

x1 = 2 + r * np.cos(theta)
y1 = r * np.sin(theta)


x2 = np.linspace(4, 6, 200)
y2 = np.zeros_like(x2)


theta2 = np.linspace(0, np.pi, 200)
x3 = 8 + r * np.cos(theta2)
y3 = 0 + r * np.sin(theta2)

# 繪圖
plt.plot(x1, y1, color="lime", linewidth=3)
plt.plot(x2, y2, color="lime", linewidth=3)
plt.plot(x3, y3, color="lime", linewidth=3)

plt.axhline(0, color="black")  # x軸
plt.axvline(0, color="black")  # y軸
plt.gca().set_aspect("equal")
plt.grid(True)
plt.show()
```
